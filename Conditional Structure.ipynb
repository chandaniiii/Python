## Nested If

```python
username = "abc@gmail.com"
password = "12345"

uname = input("Enter your username ")
pwd = input("Enter your password")

if uname!= username and pwd!= password
    print("Login error")
elif uname == username:
    if pwd == password:
        print("Your login is successful")
    else:
        print("Incorrect Password")
else:
    print("user not found")
    
```

## Match Case

```python
print("Welcome to Resturant Ordering Place")
print("1.Breakfast \n2.Lunch \n3.Dinner")
order =  int(input("Enter Your choce please"))

match order:
    case 1:
        print("You have ordered breakfast")
        print("1.Coffee and Croissant \n2.Tea and Puff")
        b_choice = int(input("Enter your breakfast choice"))
        match b_choice:
            case 1:
                print("You ordered Coffee and Croissants ")
            case 2:
                print("You ordered Tea and Puff")
    case 2:
        print("You ordered Lunch")
        print("1.Thakali Khana \n2.Mo:Mo \n3.Fried Rice")
        l_choice = int(input("Enter your lunch choice"))
        match l_choice:
            case 1:
                print("You ordered Thakali Khana ")
            case 2:
                print("You ordered Mo:Mo")
        
    case 3:
        print("You ordered Dinner")
```

## Break and Continue

```python
username = 'chand@gmail.com'
password = '123456789'

attempt = 3
for i in range(1,4):
    print(f"Remaining Attempt ={attempt}")
    uname = input("enter your username ")
    pwd = input('enter your password')

    if uname == username and pwd == password:
        print('Login Succesful')
        break
    else:
        print("Username or Password error")
        attempt = attempt - 1
        
```

### For Each Loop

```python
print('Should contain atleast 1 uppercase ')
print('Should contain atleast 1 lowercase ')
print('Should contain atleast 1 digit ')

password =  input("Enter your password ")

is_upper = False
is_lower = False
is_digit = False
is_special = False


for word in password:
    if word.isupper():
        is_upper = True
    if word.islower():
        is_lower = True
    if word.isdigit():
        is_digit = True
    if word.isdigit():
        is_digit = True
    if not word.isdigit() and not word.isupper() and not word.islower():
        is_special =  True

if is_upper and is_lower and is_digit and is_special:
    print("Login Succesful")
else:
    if is_upper == False:
        print("Should contain atleast 1 uppercase")
    if is_lower == False:
        print("Should contain atleast 1 lowercase")
    if is_digit == False:
        print("Should contain atleast 1 digit")
    if is_special == False:
        print("Should contain atleast 1 special")

```

# While Loop

### while using condition

```python
secret_num = 5
guess_num = 0

while guess_num!=secret_num:
    guess_num = int(input("Enter your guess"))
    if guess_num == secret_num:
        print("Correct")
    else:
        print("Incorrect")
    
```

### While using True but break 

```python
secret_num = 5


while True:
    guess_num = int(input("Enter your guess"))
    if guess_num == secret_num:
        print("Correct")
        break
    else:
        print("Incorrect")
```

### While Using Variable for True condition 

```python
secret_num = 5

condition = True
while condition:
    guess_num = int(input("Enter your guess"))
    if guess_num == secret_num:
        print("Correct")
        condition = False
    else:
        print("Incorrect")
```

```python
pin = 123
attempt = 3
chance = True
while chance :
    attempt!= 0
    print(f"Remaining Attempt = {attempt}")
    user_pin= int(input("Enter your pin "))
    if user_pin == pin:
        print("Correct Pin")
        chance = False
    else:
        print("Incorrect Pin")
        attempt = attempt - 1
        if attempt == 0:
            break
```
