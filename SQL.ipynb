```python
import sqlite3
```

```python
conn =sqlite3.connect('ecommerce.db')
cursor = conn.cursor()
```

## Create Table

```python
try:
    cursor.execute(
        '''
        Create Table Customers(
    	customer_id integer PRIMARY key autoincrement,
    	customer_name text not null,
    	phone text not null UNIQUE,
    	email text not null UNIQUE ,
    	address text not null,
    	customer_type text not null check(customer_type in ('Regular','Non-Regular','VIP'))
    	
    	);
        '''
    )

    print('Created table succesfully')

except sqlite3.OperationalError as e:
    print(f'Error:{e}')
```

### Staff Table

```python
try:
    cursor.execute(
        '''
        Create Table Staffs(
        staff_id integer primary key autoincrement,
        staff_name text not null,
        staff_type text not null check(staff_type in ('Full-Time','Part-Time','Contract')),
        staff_position text not null check(staff_position in ('Intern','Junior','Senior')),
        staff_email text not null unique
        )
        '''
    )
    print('Staffs Tabele Created Succesfully')

except sqlite3.OperationalError as e:
    print(f'Error as {e}')
```

### Orders Table

```python
try:
    cursor.execute('''
        Create Table Orders(
        order_id integer primary key autoincrement,
        order_status text not null check (order_status in ('Pending','Completed','Processing','Rejected')),
        order_date  date not null,
        order_desc text,
        customer_id integer not null,
        staff_id integer not null,
        foreign key(customer_id) references Customers(customer_id),
        foreign key(staff_id) references Staffs(staff_id)
                   
        );
    ''')
    print('Orders Table created successfully')

except sqlite3.OperationalError as e:
    print(f'Error: {e}')
```

## UPDATE DATA

```python
try:
    cursor.execute('''
        UPDATE Orders set order_status = 'Completed', order_date ='2026-09-22' where order_id = 2;
    
    ''')

    conn.commit()
    print('Updated values successfully')

except sqlite3.OperationalError as e:
    print(f'Error:{e}')

except sqlite3.IntegrityError as ie:
    print(f'Error:{ie}')
```

## DELETE DATA

```python
try:
    cursor.execute('''
        delete from Orders where order_id = 1
    ''')
    
    conn.commit()
    print('Delete values successfully')

except sqlite3.OperationalError as e:
    print(f'Error:{e}')

except sqlite3.IntegrityError as ie:
    print(f'Error:{ie}')
```

## SELECT Data

```python
try:
    cursor.execute('''
        select * from Customers where customer_id = 2;
    ''')

    data = cursor.fetchall()
    for row in data:
        print(row)

except OperationalError as e:
    print(f'Error:{e}')

except IntegrityError as ie:
    print(f'Error:{ie}')
```

### And , Between

```python
try:
    cursor.execute('''
        select * from Customers where customer_id > 1 and customer_id<= 4;
    ''')

    data = cursor.fetchall()
    for row in data:
        print(row)

except OperationalError as e:
    print(f'Error:{e}')

except IntegrityError as ie:
    print(f'Error:{ie}')
```
