# Error Handling 

```python
try:
    a = int(input('Enter 1st number'))
    b = int(input('Enter 2nd number'))

    c = a/b
  

except ValueError as ve:
    print('Only number can be accepted')
except ZeroDivisionError as zde:
    print('Number cant be divided by zero')

else:
    print(c)

finally:
    print('Your program run successfully ')
```

## Custom Error

```python
# 1.class
class AgeError(Exception):
    pass
```

```python
# 2.Function
def checkAge(age):
    if age>=18:
        print('You are eligible to vote')

    else:
        raise AgeError(f'Person with age {age} is not eligible to vote')
```

```python
#try & except
try:
    age = int(input('Enter your age'))
    checkAge(age)

except ValueError as ve:
    print('Only number is accepted')
except AgeError as ae:
    print(f'AgeError:{ae}')
```

```python
class PinError(Exception):
    pass

class WithdrawnAmountError(Exception):
    pass

class BalanceError(Exception):
    pass
```

```python
def atm():
    pin = "123"
    balance = 50000
    
    
    upin = input("Enter your pin")
    
    if upin == pin:
        withdraw = int(input("Enter the amount you want to withdraw "))
    
        if withdraw <= balance:
            if withdraw %1000 == 0:
             print(f"Rs{withdraw} has been withdrawn ")
            else:
                raise WithdrawnAmountError("amont must be multiple of 1000")
    
        else:
            raise BalanceError("You have less balance than the amount you want to withdraw")
            
    else:
        raise PinError("Incorrect Pin")
    
```

```python
try:
    atm()

except PinError as pe:
    print(f'Error : {pe}')
except WithdrawnAmountError as wae:
    print(f'Error : {wae}')
except BalanceError as be:
    print(f'Error : {be}')
```
