### Default Argument

```python
def signup(name='Unkown Name',contact = '1234',email = 'abc@gmail.com',password = '12345'):
    print(f'Name:{name}')
    print(f'Contact:{contact}')
    print(f'Email:{email}')
    print(f'Password:{password}')
signup('Chadani',4567892,'chada@gmail.com')
```

## All Arguments(*args)

```python
def product(*args):
    print(args)
```

```python
product('Mac',24,'$45000')  # output as a tuple 
```

```python
price1 = 9000
price2 = 3000
price3 = 94000

def changecurency(prefix, *args):
    for data in args:
        print(f'{prefix}{data}')
```

```python
changecurency('$', price1, price2, price3)
```

## ALL Keyword Arguments(**kwargs)

```python
def kw_products(**kwargs):
    for key,val in kwargs.items():
        print(f'{key}:{val}')
```

```python
kw_products(Name='Laptop',Quantity=24)  ## output as dictionary
```

 ### Lambda Function

```python
addition = lambda a,b:f'Addition of both numbers is {a+b}'
```

```python
addition(2,3)
```

```python
odd_even = lambda num:f'{num}-even'if num%2==0 else f'{num}-odd'
```

```python
odd_even(9)
```

```python
odd_even_for_at_last = lambda : [f'{num}-Even'if num%2==0 else  f'{num}-Odd' for num in range(1,21)]
```

```python
odd_even_for_at_last()
```

```python
odd_even_for_at_begining = lambda : [f'{num}-Even' for num in range(1,21) if num%2==0  ]
odd_even_for_at_begining()
```

```python
price = [400,300,600,150,200]
cheap_item=[]
expensive_item = []
```

```python
sorting_price = lambda : [cheap_item.append(item) if item<=200 else expensive_item.append(item) for item in price]

sorting_price()
```
