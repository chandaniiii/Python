# CSV File

### LIST CSV

```python
import csv
products = [
    ['Product Name','Product Quantity','Product Price','Model Year'],
    ['Mobile',2,3000,2022],
    ['TV',10,40000,2025]
]
```

```python
with open('products.csv','w',newline='') as file:
    writer = csv.writer(file)
    writer.writerows(products)
```

```python
student = [
    ['Name','Age','Grade','Fee'],
    ['Chandani',20,12,222000],
    ['Site',21,12,456789]
]
with open('students.csv','w',newline='') as f:
    content = csv.writer(f)
    content.writerows(student)
```

### Read File

```python
with open('students.csv','r') as file:
    content = csv.reader(file)
    for row in content:
        print(row)
```

```python
new_value = [
    ['Ram',21,13,234567],
    ['Shyam',20,14,7890876]
]
with open('students.csv','a',newline='') as myfile:
    content = csv.writer(myfile)
    content.writerows(new_value)
```

## Dictionary CSV

```python
Product_j = [
    {
        'Product Name':'Mac',
        'Product Brand':'Apple',
        'Product Cost':'$120'
    },
    {
        'Product Name':'Charger',
        'Product Brand':'Anker',
        'Product Cost':'$160'
        
    },
    {
        'Product Name':'Phone',
        'Product Brand':'Samsung',
        'Product Cost':'$20000'
    }
]
Product_j
```

```python
# Write to file

import csv
with open('products_dict.csv','w',newline='') as file:
    columns = ['Product Name','Product Brand','Product Cost']
    content = csv.DictWriter(file,fieldnames = columns)
    content.writeheader()
    content.writerows(Product_j)
    
```

```python
#Read File

with open('products_dict.csv','r') as f:
    reader = csv.DictReader(f)
    for row in reader:
        for key,val in row.items():
            print(f'{key}:{val}')
        print('\n')
```

```python
new_product_j =[{
    'Product Name' : 'Charger',
    'Product Brand' : 'HP',
    'Product Cost' : '$1500'
}]
```
